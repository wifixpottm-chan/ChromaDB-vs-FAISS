```python
from bs4 import BeautifulSoup
import requests
import json
```

```python
Urls = ["https://myanimelist.net/anime/20/Naruto/episode",
"https://myanimelist.net/anime/20/Naruto/episode?offset=100",
"https://myanimelist.net/anime/20/Naruto/episode?offset=200"]
```

```python
links = []
for urls in Urls:

    main = requests.get(urls)
    soup1 = BeautifulSoup(main.content, 'html.parser')

    for tr in soup1.find_all('tr', class_='episode-list-data'):
        a_tag = tr.find('a')
        if a_tag:
            links.append(a_tag['href'])
print(links)
print(len(links))
```

```python
dataset = []
for i in links:
    main = requests.get(i)
    soup2 = BeautifulSoup(main.content, 'html.parser')

    summary_div = soup2.find("div", class_="pt8 pb8")

    if summary_div is None:
        print("Summary not found:", i)
        continue

    summary = summary_div.get_text(" ", strip=True)

    dataset.append({
        "id": len(dataset) + 1,
        "url": i,
        "raw_text": summary
    })
```

```python
with open("naruto_s1.json", "w", encoding="utf-8") as f:
    json.dump(dataset, f, ensure_ascii=False, indent=4)

print("Saved", len(dataset), "articles")
```
